## 1) Puntuación del campeonato de fútbol


Nuestro equipo de fútbol terminó el campeonato.
Los resultados de los partidos están registrados en una colección de *strings*. Cada partido está representado por un texto con el formato `"x:y"`, donde `x` es el marcador de **nuestro** equipo y `y` el marcador del **rival**.

**Ejemplo:** `["3:1", "2:2", "0:1", ...]`

La asignación de puntos por partido es:

- si `x > y`: **3 puntos** (victoria)
- si `x < y`: **0 puntos** (derrota)
- si `x = y`: **1 punto** (empate)

Escribir una función que tome esta colección y devuelva el **total de puntos** que obtuvo nuestro equipo (`x`) en el campeonato según las reglas indicadas.

**Notas**

- nuestro equipo siempre juega **10** partidos en el campeonato
- `0 <= x <= 4`
- `0 <= y <= 4`

**Ejemplos:**

```python
# Ejemplo 1
Input: ["3:1", "2:2", "0:1", "4:0", "3:3", "1:0", "2:1", "3:1", "1:1", "0:2"]
Output: 18
# Explicación: victoria(3) + empate(1) + derrota(0) + victoria(3) + empate(1) + victoria(3) + victoria(3) + victoria(3) + empate(1) + derrota(0) = 18

# Ejemplo 2
Input: ["1:0", "2:0", "3:0", "4:0", "2:1", "3:1", "4:1", "3:2", "4:2", "4:3"]
Output: 30
# Explicación: 10 victorias = 10 × 3 = 30 puntos
```

````python
resultado = "4:1"
partido = resultado.split(":")

print(partido[0], partido[1])

````

````python
def calcular_puntaje(resultados):
    # sabemos que resultados es una lista con formato "x:y"

    puntaje = 0

    valores = []

    for resultado in resultados:
        resultado_partido = resultado.split(":")  # x:y -> [x, y]
        resultado_nuestro = int(resultado_partido[0])  # tomamos x -> '2'
        resultado_rival = int(resultado_partido[1])  # tomamos y -> '3'
        # resultado_nuestro, resultado_rival = resultado.split(":")

        # print(
        #     f"resultado_nuestro={resultado_nuestro}, resultado_rival={resultado_rival}"
        # )

        if resultado_nuestro > resultado_rival:
            # ganamos
            puntaje += 3
            valores.append(3)
        elif resultado_rival > resultado_nuestro:
            # perdimos
            puntaje += 0
            valores.append(0)
        else:  # es lo mismo q usar un elif resultado_rival == resultado_nuestro
            # empate
            puntaje += 1
            valores.append(1)

    return [puntaje, sum(valores)]
````

````python
resultados = ["3:1", "2:2", "0:1", "4:0", "3:3", "1:0", "2:1", "3:1", "1:1", "0:2"]
calcular_puntaje(resultados)
````

````python
resultados2 = ["1:0", "2:0", "3:0", "4:0", "2:1", "3:1", "4:1", "3:2", "4:2", "4:3"]
calcular_puntaje(resultados2)
````
